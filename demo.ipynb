# Stock Lakehouse Demo
## Doc data that tu MinIO (Bronze Layer)


```python
import pandas as pd
import pyarrow.parquet as pq
import pyarrow.fs as fs

# Ket noi MinIO - dung host.docker.internal de truy cap tu container
s3_fs = fs.S3FileSystem(
    endpoint_override='http://host.docker.internal:9000',
    access_key='minioadmin',
    secret_key='minioadmin',
    scheme='http'
)

# Doc FPT data tu Bronze layer
path = 'stock-bronze/symbol=FPT/year=2026/month=09/part-001.parquet'
table = pq.read_table(path, filesystem=s3_fs)
df = table.to_pandas()

print(f'Table: stock-bronze/symbol=FPT')
print(f'Records: {len(df)}')
df.head()
```

```python
import matplotlib.pyplot as plt

df['timestamp'] = pd.to_datetime(df['timestamp'])
plt.figure(figsize=(12, 6))
plt.plot(df['timestamp'], df['close'], color='blue')
plt.title('FPT Stock Price - September 2026')
plt.xlabel('Date')
plt.ylabel('Price (VND)')
plt.grid(True)
plt.show()
```
